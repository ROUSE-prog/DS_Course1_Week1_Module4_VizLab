## Exercise 1 — Create a Bar Chart

### Business Scenario
Jim's Video Library contains 40 crime, 30 science fiction, 10 drama, 50 comedy, 25 action, and 5 documentary movies.

### Task
Create a vertical bar chart using `ax.bar()` that:

* hard-codes the six category counts
* uses six evenly spaced x positions
* displays the genre names using `tick_label`
* includes an informative title and axis labels

The graph should communicate the composition of the store's movie inventory at a glance.


```python
# CodeGrade step1

# Hard-code the movie counts and genre labels
height = [40, 30, 10, 50, 25, 5]
x = range(6)
labels = ["Crime", "Sci/fi", "Drama", "Comedy", "Action", "Documentary"]

# Create the plot
fig, ax = plt.subplots(figsize=(8, 6))

# Plot vertical bars
bar_chart = ax.bar(x, height, tick_label=labels)

# Give a title to the bar graph and label the axes
ax.set_title("Jim's Video Library")
ax.set_xlabel("Genres")
ax.set_ylabel("Number of Movies")
```

### Analyst Interpretation
Comedy is the largest category with **50 movies**, followed by crime with **40**. Documentary is the smallest category with only **5 movies**. A bar chart is appropriate because the goal is to compare counts across discrete categories.


## Exercise 2 — Create a Scatter Plot

### Business Scenario
You are a data scientist working for a major car manufacturer. The company wants to understand how vehicle weight relates to fuel efficiency.

### Task
Create a scatter plot using `ax.scatter()` with:

* car weight on the x-axis
* miles per gallon on the y-axis
* a descriptive title
* clear axis labels
* a legend describing the plotted relationship

Use the supplied values exactly as provided in the lab.


```python
# CodeGrade Step2

weight = [2750, 3125, 2100, 4082, 2690, 3640, 4380, 2241, 2895, 3659]
mpg = [29, 23, 33, 28, 20, 21, 14, 25, 31, 17]

# Create the plot objects fig and ax
fig, ax = plt.subplots(figsize=(8, 6))

# Plot with scatter()
scatter_plot = ax.scatter(weight, mpg, label="weight vs. mileage")

# Set x and y axes labels, legend, and title
ax.set_xlabel("Car Weight")
ax.set_ylabel("Miles per Gallon")
ax.legend()
ax.set_title("Consumer Cars")
```

### Analyst Interpretation
The scatter plot suggests a **negative relationship** between car weight and fuel efficiency: as vehicle weight increases, MPG generally decreases. That pattern is useful to an automotive team because it highlights weight reduction as a possible lever for improving fuel economy.


## Exercise 3 — Create a Histogram

### Business Scenario
Joe manages a bank branch and has received complaints about long customer wait times. He records the waiting time, in seconds, for 20 customers.

### Task
Create a histogram using `ax.hist()` that:

* uses the provided waiting-time observations
* uses `bins=5`
* labels both axes clearly
* uses the title shown in the reference visualization

The histogram should help Joe understand the distribution of customer waiting times rather than only individual observations.


```python
# CodeGrade Step3

x = [43.1, 35.6, 37.5, 36.5, 45.3, 43.4,
     40.3, 50.2, 47.3, 31.2, 42.2, 45.5,
     30.3, 31.4, 35.6, 45.2, 54.1, 45.6,
     36.5, 43.1]

# Create the plot objects fig and ax
fig, ax = plt.subplots(figsize=(8, 6))

# Plot the histogram with hist() function
my_hist = ax.hist(x, bins=5)

# Label axes and set title
ax.set_xlabel("Waiting Time")
ax.set_ylabel("Number of Customers")
ax.set_title("Customer Waiting Times")
```

### Analyst Interpretation
Most observed waits fall in the middle of the distribution, roughly from the mid-30s through the upper-40s seconds, while relatively few customers experienced the shortest or longest waits. The histogram gives Joe a clearer picture of the overall customer experience than a raw list of times alone.
